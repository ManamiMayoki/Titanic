```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math
```

```python
# Geometric Rotation
x = [1,4,4,1,1]
y = [1,1,3,3,1]
theta = math.pi/4
xr = [x[i]*math.cos(theta) - y[i]*math.sin(theta) for i in range(len(x))]
yr = [x[i]*math.sin(theta) + y[i]*math.cos(theta) for i in range(len(x))]
plt.plot(x, y, marker='o')
plt.plot(xr, yr, marker='o')
plt.gca().set_aspect('equal')
plt.grid(True)
plt.show()
```

```python
#Coordinate Rotation
x = [1, 4, 4, 1, 1]
y = [1, 1, 3, 3, 1]
theta = math.pi/4
xc = [x[i]*math.cos(theta) + y[i]*math.sin(theta) for i in range(len(x))]
yc = [-x[i]*math.sin(theta) + y[i]*math.cos(theta) for i in range(len(x))]
plt.plot(x, y, marker='o', label='Original (Global) Frame')
plt.plot(xc, yc, marker='o', label='New (Local) Frame View')
plt.gca().set_aspect('equal')
plt.grid(True)
plt.legend()
plt.show()
```

```python
#Scaling
x = [1,4,4,1,1]
y = [1,1,3,3,1]
sx = 2
sy = 2
xs = [sx*i for i in x]
ys = [sy*i for i in y]
plt.plot(x, y, marker='o')
plt.plot(xs, ys, marker='o')
plt.gca().set_aspect('equal')
plt.grid(True)
plt.show()
```

```python
#Coordinate Scaling
x = [1, 4, 4, 1, 1]
y = [1, 1, 3, 3, 1]

sx = 2
sy = 2

xc = [i / sx for i in x]
yc = [i / sy for i in y]

plt.plot(x, y, marker='o', label='Original Coordinates')
plt.plot(xc, yc, marker='o', label='New Coordinate View')
plt.gca().set_aspect('equal')
plt.grid(True)
plt.legend()
plt.show()
```
