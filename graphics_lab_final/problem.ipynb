```python
#DDA

import matplotlib.pyplot as plt
X1, Y1 = 2, 3
X2, Y2 = 12, 8
DX = X2 - X1
DY = Y2 - Y1
STEPS = max(abs(DX), abs(DY))
X_INC = DX / STEPS
Y_INC = DY / STEPS
X, Y = X1, Y1
plt.figure()
plt.grid(True)
plt.axis('equal')
plt.title('DDA Algorithm')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
X_VALS, Y_VALS = [], []
for I in range(STEPS + 1):
    X_VALS.append(round(X))
    Y_VALS.append(round(Y))
    X += X_INC
    Y += Y_INC
plt.plot(X_VALS, Y_VALS, 'ks', markersize=8, markerfacecolor='k')
plt.plot([X1, X2], [Y1, Y2], 'r--')
plt.show()
```

```python
#Flood Fill Algorithm

import numpy as np
import matplotlib.pyplot as plt
import sys
sys.setrecursionlimit(10000)
canvas = np.zeros((100,100))
canvas[20:80,20] = 1
canvas[20:80,80] = 1
canvas[20,20:80] = 1
canvas[80,20:80] = 1
def flood_fill(x,y,fill,old):
    if canvas[x][y] == old:
        canvas[x][y] = fill
        flood_fill(x+1,y,fill,old)
        flood_fill(x-1,y,fill,old)
        flood_fill(x,y+1,fill,old)
        flood_fill(x,y-1,fill,old)
flood_fill(50,50,2,0)
plt.imshow(canvas)
plt.title("Flood Fill Output")
plt.show()
```

```python
import matplotlib.pyplot as plt

def bresenham(x1, y1, x2, y2):
    """Generates the pixels/points between two coordinates using Bresenham's algorithm."""
    points = []
    dx = abs(x2 - x1)
    dy = abs(y2 - y1)
    
    sx = 1 if x1 < x2 else -1
    sy = 1 if y1 < y2 else -1
    err = dx - dy
    
    while True:
        points.append((x1, y1))
        if x1 == x2 and y1 == y2:
            break           
        e2 = 2 * err
        if e2 > -dy:
            err -= dy
            x1 += sx
        if e2 < dx:
            err += dx
            y1 += sy 
    return points
top = (50, 90)
bottom = (50, 40)
left = (20, 70)
right = (80, 70)

tail_left = (40, 20)
tail_right = (60, 20)

lines_to_draw = [
    (top, right),
    (right, bottom),
    (bottom, left),
    (left, top),
    (top, bottom),
    (left, right),
    

    (bottom, tail_left),   
    (bottom, tail_right), 
    (tail_left, tail_right)
]


all_kite_points = []
for start, end in lines_to_draw:
    line_points = bresenham(start[0], start[1], end[0], end[1])
    all_kite_points.extend(line_points)

x_coords = [p[0] for p in all_kite_points]
y_coords = [p[1] for p in all_kite_points]


plt.figure(figsize=(6, 8))
plt.scatter(x_coords, y_coords, color='darkcyan', s=5, label="Bresenham Pixels")


plt.title("Kite with a Geometric Triangular Tail", fontsize=14)
plt.xlabel("X Coordinate")
plt.ylabel("Y Coordinate")
plt.grid(True, which='both', linestyle='--', color='lightgray')
plt.axis('equal')
plt.legend()


plt.show()
```
